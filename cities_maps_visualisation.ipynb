```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
small_set_20 = pd.read_csv('data/cities_20.csv')
medium_set_50 = pd.read_csv('data/cities_50.csv')
large_set_100 = pd.read_csv('data/cities_100.csv')
```

```python
def geoplot_cities(df, title="Mapa miast"):
    fig, ax = plt.subplots(figsize=(8, 10))

    ax.scatter(df['longitude'], df['latitude'],
               s=60, color='red', alpha=0.5, zorder=5)

    
    ax.set_title(title)
    ax.set_xlabel('Długość geograficzna')
    ax.set_ylabel('Szerokość geograficzna')
    ax.grid(True, linestyle='--', alpha=0.4)
    plt.tight_layout()
    plt.show()
```

```python
geoplot_cities(small_set_20, title="20 największych miast Polski")
geoplot_cities(medium_set_50, title="50 największych miast Polski")
geoplot_cities(large_set_100, title="100 największych miast Polski")
```
